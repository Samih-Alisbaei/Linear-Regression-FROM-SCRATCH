```python
#this imports the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv(r"C:\Users\zezoo\Downloads\train\Salary Data.csv")
data.head()
```

```python
X_train=data["YearsExperience"].values
y_train=data["Salary"].values
```

```python
plt.scatter(x=X_train,y=y_train,colorizer="blue")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show
```

```python
def cost_function(x,y,w,b):
    #y=mx+b so m is the number of itreations 
    m=len(x)
    mean_square_error=0
    for i in range(m):
        fun=(w*x[i])+b
        cost=(y[i]-fun)**2
        mean_square_error+=cost
    return (1/2*m)*mean_square_error
```

```python
def gradient_function(x,y,w,b):
    ini_dw=0
    ini_db=0
    m=len(x)
    for i in range(m):
        fun=(w*x[i])+b
        ini_dw+=(fun-y[i])*x[i]
        ini_db+=(fun-y[i])
    final_dw=(1/m)*ini_dw
    final_db=(1/m)*ini_db
    return final_dw , final_db
```

```python
def gradient_descent(x,y,alpha,loops):
    w=0
    b=0
    for i in range(loops):
        dw,db=gradient_function(x,y,w,b)

        w=w-alpha*dw
        b=b-alpha*db

        print(f"itreation {i} , Cost : {cost_function(x,y,w,b)} ")
    return w,b
```

```python


w_final, b_final = gradient_descent(X_train, y_train, 0.01, 300)


X_train = np.array(X_train)
y_pred = w_final * X_train + b_final

plt.figure(figsize=(8, 5))

plt.scatter(X_train, y_train, color="blue", label="Actual Data", alpha=0.6)

plt.plot(X_train, y_pred, color="red", linewidth=2, label="Linear Regression Line")

plt.xlabel("X (Feature)")
plt.ylabel("y (Target)")
plt.title("Linear Regression Line Fit")
plt.legend()
plt.grid(True)

plt.show()
```
